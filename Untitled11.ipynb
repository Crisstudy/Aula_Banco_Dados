```r
# Instale a biblioteca neuralnet
install.packages ("neuralnet")
```

```r
library (neuralnet)

#dados ficticios de treinamento
set.seed(123) #decide a semente para reprodução dos resultados
x_train <- data.frame(Input1 = runif(100, 0, 1), Input2 = runif(100, 0, 1))
y_train <- x_train$Input1 + x_train$Input2 + rnorm(100, mean = 0, sd = 0.1)

#dataframe que combina os recursos de entrada e o alvo
train_data <- data.frame(x_train, y_train)

#treinando uma rede neural
model <- neuralnet(
  y_train ~ Input1 + Input2,
  data = train_data,
  hidden = c(5, 3), #Camadas ocultas com 5 e 3 neurônios
  linear.output = TRUE # saída linear para regressão
)

# Visualizar a rede neural
plot(model)

# fazer previsões com a rede neural
new_data <- data.frame(Input1 = 0.6, Input2 = 0.4) # dados de entrada para revisão
predictions <- predict(model, newdata = new_data)
cat('Previsão:', predictions, '\n')
```
